```python
m = sales[sales['Product'] == 'Mint Chip Choco']
d = sales[sales['Product'] == '85% Dark Bars']
p = sales[sales['Product'] == 'Peanut Butter Cubes']
```

```python
m = m.groupby('Date')['Amount'].sum()
d = d.groupby('Date')['Amount'].sum()
p = p.groupby('Date')['Amount'].sum()
```

```python
m.index = pd.CategoricalIndex(m.index, categories=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August'],ordered=True)
d.index = pd.CategoricalIndex(d.index, categories=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August'],ordered=True)
p.index = pd.CategoricalIndex(p.index, categories=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August'],ordered=True)
```

```python
m=m.sort_index()
d=d.sort_index()
p=p.sort_index()
```

```python
d
```

```python
x=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August']
```

```python
plt.figure(figsize=(15, 7))
plt.title(' ')
plt.xlabel('Month')
plt.ylabel('Total Amount')
plt.bar(np.arange(m.shape[0]) - 0.2, m.values, color='#250E75', width=0.2, label='Mint Chip Choco')
plt.bar(np.arange(d.shape[0]), d.values, color='#8B1C62', width=0.2, label='85% Dark Bars')
plt.bar(np.arange(p.shape[0]) + 0.2, p.values, color='#0A9975', width=0.2, label='Peanut Butter Cubes')
plt.xticks(np.arange(m.shape[0]) - 0.2, x)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(15, 7))
plt.title(' ')
plt.xlabel('Month')
plt.ylabel('Total Amount')
plt.bar(m.index, m.values, color='#250E75', width=0.2, label='Mint Chip Choco',)
plt.bar(d.index, d.values, color='#8B1C62', width=0.2, label='85% Dark Bars',bottom=m.values)
plt.bar(p.index, p.values, color='#0A9975', width=0.2, label='Peanut Butter Cubes',bottom=m.values+d.values)
plt.legend()
plt.show()
```
